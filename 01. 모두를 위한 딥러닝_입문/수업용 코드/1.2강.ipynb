조건문과 반복문

```python
# 조건문
score = 85

if score >= 90:
    print("A 학점")
elif score >= 80:
    print("B 학점")
else:
    print("C 학점")

# 반복문 (1부터 5까지 출력)
for i in range(1, 6):
    print(i)
```

함수 및 모듈

```python
# 함수 정의
def greet(name):
    return "안녕하세요, " + name + "님!"

# 함수 호출
message = greet("Kim")
print(message)

# 모듈 사용
import math
print("원주율:", math.pi)
```

Matrix 연산

```python
import numpy as np

# 행렬(matrix) 생성
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

result_addition = A + B
print("덧셈 결과:\n", result_addition)

result_subtraction = A - B
print("뺄셈 결과:\n", result_subtraction)
```

곱셈 연산

```python
result_elementwise = A * B
print("요소별 곱셈 결과:\n", result_elementwise)

result_dot = np.dot(A, B)
print("내적 결과:\n", result_dot)
# 또는
result_dot_alternative = A @ B
print("내적 결과 (alternative):\n", result_dot_alternative)
```
